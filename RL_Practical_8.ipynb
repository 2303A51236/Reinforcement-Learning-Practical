```python
# Assignment 8 - Reinforcement Learning
# Calculate State Value and Action Value
# Without using Bellman Equation

GAMMA = 0.9

# Grid:
# S0  S1  G
# X   S3  S4

states = ["S0", "S1", "S3", "S4", "G"]

# Policy given in the assignment
policy = {
    "S0": "right",
    "S1": "down",
    "S3": "right",
    "S4": "up"
}

# Rewards
NORMAL_REWARD = -1
GOAL_REWARD = 10
OBSTACLE_REWARD = -5


# ---------------------------------------------------
# Environment transition function
# ---------------------------------------------------
def step(state, action):

    transitions = {
        "S0": {
            "up": "S0",
            "down": "S0",       # obstacle X, so stays
            "left": "S0",
            "right": "S1"
        },

        "S1": {
            "up": "S1",
            "down": "S3",
            "left": "S0",
            "right": "G"
        },

        "S3": {
            "up": "S1",
            "down": "S3",
            "left": "X",
            "right": "S4"
        },

        "S4": {
            "up": "G",
            "down": "S4",
            "left": "S3",
            "right": "S4"
        }
    }

    next_state = transitions[state][action]

    # Goal
    if next_state == "G":
        return next_state, GOAL_REWARD

    # Obstacle
    if next_state == "X":
        return next_state, OBSTACLE_REWARD

    # Normal state
    return next_state, NORMAL_REWARD


# ---------------------------------------------------
# State Value Function
# ---------------------------------------------------
def state_value_function(start_state):

    state = start_state
    discount = 1
    total_return = 0

    while state != "G":

        action = policy[state]

        next_state, reward = step(state, action)

        total_return += discount * reward

        discount *= GAMMA

        state = next_state

    return total_return


# ---------------------------------------------------
# Action Value Function
# ---------------------------------------------------
def action_value_function(start_state, action):

    # Take the specified action first
    next_state, reward = step(start_state, action)

    total_return = reward
    discount = GAMMA

    # Continue following the policy
    state = next_state

    while state != "G":

        policy_action = policy[state]

        next_state, reward = step(state, policy_action)

        total_return += discount * reward

        discount *= GAMMA

        state = next_state

    return total_return


# ---------------------------------------------------
# Main
# ---------------------------------------------------

# State value of S0
V_S0 = state_value_function("S0")

# Action value of S0 when taking UP
Q_S0_UP = action_value_function("S0", "up")

print("State Value Function")
print("--------------------")
print("V(S0) =", V_S0)

print("\nAction Value Function")
print("---------------------")
print("Q(S0, UP) =", Q_S0_UP)
```
