```python
import gymnasium as gym


def run_random_agent(environment_name, episodes=5):

    # Create environment
    env = gym.make(environment_name)

    print("\n===================================")
    print("Environment:", environment_name)
    print("===================================")

    episode_rewards = []

    for episode in range(1, episodes + 1):

        # Reset environment
        state, info = env.reset()

        total_reward = 0
        steps = 0

        terminated = False
        truncated = False

        print("\nEpisode:", episode)
        print("Initial State:", state)

        while not terminated and not truncated:

            # Select a random action
            action = env.action_space.sample()

            # Perform action
            next_state, reward, terminated, truncated, info = env.step(action)

            # Update values
            state = next_state
            total_reward += reward
            steps += 1

        # Calculate average reward
        average_reward = total_reward / steps

        episode_rewards.append(total_reward)

        print("Steps:", steps)
        print("Total Reward:", total_reward)
        print("Average Reward:", round(average_reward, 2))

    # Overall average
    overall_average = sum(episode_rewards) / len(episode_rewards)

    print("\n===================================")
    print("FINAL RESULTS")
    print("===================================")
    print("Episode Rewards:", episode_rewards)
    print("Average Total Reward:", round(overall_average, 2))

    env.close()


# =========================================
# 1. CLIFF WALKING
# =========================================

print("\n===== CLIFF WALKING =====")

run_random_agent("CliffWalking-v1", episodes=5)


# =========================================
# 2. TAXI
# =========================================

print("\n===== TAXI =====")

run_random_agent("Taxi-v4", episodes=5)
```
