```python
import random

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

# Part B – Initialize N, R and Q tables
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0, 'B': 0, 'C': 0}

# Given reward patterns
rewards = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Keep track of the next reward for each machine
reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}

# Parameters
N_play = 10
epsilon = 0.2

total_reward = 0

print("MODIFIED ε-GREEDY STRATEGY")
print("-" * 50)

# Part C – Modified ε-Greedy Decision
for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    # Find highest Q-value
    max_Q = max(Q.values())

    # Find machines with highest Q-value
    best_machines = [
        machine for machine in machines
        if Q[machine] == max_Q
    ]

    # Exploration
    if r < epsilon:

        # If all machines have the same Q-value,
        # choose any machine randomly
        if len(best_machines) == len(machines):
            selected_machine = random.choice(machines)

        else:
            # Select randomly except the best machine
            other_machines = [
                machine for machine in machines
                if machine not in best_machines
            ]

            selected_machine = random.choice(other_machines)

        decision = "Exploration"

    # Exploitation
    else:
        selected_machine = random.choice(best_machines)
        decision = "Exploitation"

    # Part D – Reward Collection

    # Get reward from the given reward pattern
    index = reward_index[selected_machine] % len(
        rewards[selected_machine]
    )

    reward = rewards[selected_machine][index]

    # Move to next reward
    reward_index[selected_machine] += 1

    # Part E – Q-Value Update

    # Increase participation count
    N[selected_machine] += 1

    # Add reward
    R[selected_machine] += reward

    # Incremental sample-average formula
    Q[selected_machine] = (
        Q[selected_machine]
        + (1 / N[selected_machine])
        * (reward - Q[selected_machine])
    )

    # Add to total reward
    total_reward += reward

    # Display play details
    print("\nPlay:", play)
    print("Random Number:", round(r, 4))
    print("Decision:", decision)
    print("Selected Machine:", selected_machine)
    print("Reward:", reward)

    print("N:", N)
    print("R:", R)

    print(
        "Q:",
        {machine: round(Q[machine], 2) for machine in machines}
    )


# Part F – Final Results

print("\n" + "=" * 65)
print("FINAL RESULTS")
print("=" * 65)

print(
    "\nMachine\tN (Count)\tR (Total Reward)\tQ (Estimated Reward)"
)

print("-" * 65)

for machine in machines:
    print(
        machine,
        "\t", N[machine],
        "\t\t", R[machine],
        "\t\t", round(Q[machine], 2)
    )

print("\nTotal Reward received by player:", total_reward)

print("\nAnalysis:")

for machine in machines:
    print(
        "Machine", machine,
        "was selected", N[machine],
        "times, received total reward", R[machine],
        "and has final Q-value", round(Q[machine], 2)
    )
```
