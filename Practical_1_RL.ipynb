```python
n = int(input("Enter number of machines: "))

machines = {}
q = {}

for i in range(n):
    name = input("Enter machine name: ")
    rewards = list(map(int, input("Enter rewards separated by space: ").split()))

    machines[name] = rewards
    q[name] = sum(rewards) / len(rewards)

for name in machines:
    print("\nMachine", name)
    print("Rewards:", machines[name])
    print("Q(" + name + ") =", q[name])

selected = max(q, key=q.get)

print("\nGreedy Selected Machine:", selected)
```

```python
n = int(input("Enter number of machines: "))

q = {}
count = {}

for i in range(n):
    name = input("Enter machine name: ")
    rewards = list(map(int, input("Enter rewards: ").split()))

    q[name] = 0
    count[name] = 0

    for reward in rewards:
        count[name] = count[name] + 1

        q[name] = q[name] + (reward - q[name]) / count[name]

        print("\nMachine:", name)
        print("Reward:", reward)
        print("N =", count[name])
        print("Q =", q[name])

print("\nFinal Q Values")

for name in q:
    print("Q(" + name + ") =", q[name])

selected = max(q, key=q.get)

print("\nGreedy Selected Machine:", selected)
```
